```python
subject_marks = np.array([                  #Create a NumPy array
    math,
    science,
    english
])

# print(subject_marks)
```

```python
subject_average = np.mean(subject_marks, axis=1)               #Calculate subject averages

# print("Math Average:", subject_average[0])

# print("Science Average:", subject_average[1])

# print("English Average:", subject_average[2])
```

```python
best_subject = np.argmax(subject_average)                   #Find the best-performing subject

subjects = np.array(["Math", "Science", "English"])

print("Best Performing Subject:", subjects[best_subject])
```
